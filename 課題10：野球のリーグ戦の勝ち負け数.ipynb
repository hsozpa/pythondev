```python
class Baseball_Team:
    def __init__(self, name: str, win: int = 0, lose: int = 0, draw: int = 0):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    def calc_win_rate(self):
        total = self.win + self.lose
        return self.win / total
    

    def show_team_result(self):
        return f"{self.name:<9}{self.win:>4}{self.lose:>5}{self.draw:>5} {self.calc_win_rate():>5.3f}"
```

```python
teams = [
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Meteors", 71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks", 70, 70, 3),
    Baseball_Team("Pythons", 68, 73, 2),
    Baseball_Team("Owls", 59, 82, 2),
]


print(f"{'team':<9}{'win':>4}{'lose':>5}{'draw':>5} {'rate':>5}")

for i in teams:
    print(i.show_team_result())
```
