```python
# --- 1. 割られる数の入力 ---
a_input = input("割られる数を入力してください：")

try:
    a = int(a_input)

except ValueError:
    print(f"割られる数を入力してください：{a_input}")
    print("エラー：数値を入力してください")
    print("処理を終了します")
    raise SystemExit        # ここでプログラムを終了


# --- 2. 割る数の入力（ここに来るのは a が整数に変換できたときだけ） ---
b_input = input("割る数を入力してください：")

try:
    b = int(b_input)
    c = a / b    

except ZeroDivisionError:
    print(f"割る数を入力してください：{a_input}")
    print(f"割る数を入力してください：{b_input}")
    print("エラー：0で割り算しないでください")

except ValueError:
    print(f"割られる数を入力してください：{b_input}")
    print("エラー：数値を入力してください")
    raise SystemExit        # ここでプログラムを終了

else:
    print(f"割られる数を入力してください：{a}")
    print(f"割る数を入力してください：{b}")
    print(f"{a} ÷ {b} = {c}")

finally:
    print("処理を終了します")
```
