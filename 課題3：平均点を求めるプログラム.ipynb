```python
total_score = 0
count = 0

while True:
    score = int(input("点数を入力してください："))
    print(f"点数を入力してください：{score}")

    if score == -1:
        break

    total_score += score
    count += 1

average = total_score / count

print(f"{count}人のテストの平均点は{average:.1f}点です")
```
