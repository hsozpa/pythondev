```python
words = []

alphabet_count = dict.fromkeys(list("abcdefghijklmnopqrstuvxyz"), 0)


while True:
    word = input("英単語を入力してください：")
    print(f"英単語を入力してください：{word}")

    if word == "":
        break

    words.append(word)


    for char in word:
        if char in alphabet_count:
            alphabet_count[char] += 1

words.sort()


print("入力した英単語：", words) 
for letter, count in alphabet_count.items():
    if count > 0:
        print(f"{letter}が{count}個ありました")
```
